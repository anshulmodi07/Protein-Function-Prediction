```python
"""
Improved Protein Function Prediction Pipeline
Computational Biology & PyTorch Implementation
Fixes:
1. Data leakage in label filtering and splitting
2. Padding contamination in Conv1D / AdaptiveMaxPool1d via masked pooling
3. Severe class imbalance via smoothed positive class weighting (pos_weight)
4. Validation-based threshold tuning (never tuning on test set)
5. Multi-scale residual CNN architecture (kernels 3, 5, 9, 15 + BatchNorm + Dual Pooling)
6. Early stopping + Learning rate scheduling
"""

import os
import re
import random
import time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.metrics import f1_score, precision_score, recall_score
from torch.utils.data import Dataset, DataLoader
```

```python
# ==================================================
# 1. DETERMINISTIC REPRODUCIBILITY
# ==================================================
SEED = 42
def seed_everything(seed=SEED):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True

seed_everything(SEED)

# ==================================================
# 2. CONFIGURATION
# ==================================================
MAX_LEN = 1000
MIN_LABEL_COUNT = 25
MAX_LABELS = 40
BATCH_SIZE = 32
EPOCHS = 15
LR = 0.001

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Executing on Device: {DEVICE}")

AMINO_ACIDS = "ACDEFGHIKLMNPQRSTVWY"
VOCAB = {aa: i + 1 for i, aa in enumerate(AMINO_ACIDS)}
PAD_IDX = 0
```

```python
# ==================================================
# 3. DATA LOADING & LEAK-FREE PREPROCESSING
# ==================================================
if not os.path.exists("uniprot_proteins.csv"):
    raise FileNotFoundError("uniprot_proteins.csv is required in the working directory.")

df = pd.read_csv("uniprot_proteins.csv")
df = df.dropna(subset=["sequence", "go_terms"]).copy()
df["sequence"] = df["sequence"].astype(str).str.upper().str.strip()
df = df[df["sequence"].str.fullmatch(r"[ACDEFGHIKLMNPQRSTVWY]+")].copy()

def extract_go_terms(val):
    return list(set(re.findall(r"GO:\d{7}", str(val))))

df["labels"] = df["go_terms"].apply(extract_go_terms)
df = df[df["labels"].map(len) > 0]

all_sequences = df["sequence"].tolist()
all_labels = df["labels"].tolist()

# To prevent data leakage, we establish the train/val/test splits first!
indices = np.arange(len(all_sequences))
train_idx_raw, temp_idx = train_test_split(indices, test_size=0.30, random_state=SEED)
val_idx_raw, test_idx_raw = train_test_split(temp_idx, test_size=0.50, random_state=SEED)

# Determine selected labels STRICTLY based on training set frequencies!
train_label_counter = Counter(term for i in train_idx_raw for term in set(all_labels[i]))
eligible_labels = {term for term, count in train_label_counter.items() if count >= MIN_LABEL_COUNT}
selected_labels = sorted(
    eligible_labels,
    key=lambda term: train_label_counter[term],
    reverse=True
)[:MAX_LABELS]

print(f"Selected {len(selected_labels)} GO terms derived strictly from training set statistics.")

# Filter labels to selected vocabulary
filtered_labels = [
    [term for term in row if term in set(selected_labels)]
    for row in all_labels
]

# Keep proteins with at least one target label
valid_mask = [len(row) > 0 for row in filtered_labels]
valid_indices = set([i for i, v in enumerate(valid_mask) if v])

train_idx = [i for i in train_idx_raw if i in valid_indices]
val_idx = [i for i in val_idx_raw if i in valid_indices]
test_idx = [i for i in test_idx_raw if i in valid_indices]

mlb = MultiLabelBinarizer(classes=selected_labels)
Y_all = mlb.fit_transform(filtered_labels).astype(np.float32)

def encode_seq(seq, max_len=MAX_LEN):
    seq = seq[:max_len]
    arr = [VOCAB[aa] for aa in seq]
    arr += [PAD_IDX] * (max_len - len(arr))
    return arr

X_all = np.array([encode_seq(seq) for seq in all_sequences], dtype=np.int64)

X_train, Y_train = X_all[train_idx], Y_all[train_idx]
X_val, Y_val = X_all[val_idx], Y_all[val_idx]
X_test, Y_test = X_all[test_idx], Y_all[test_idx]

print(f"Clean Partition Sizes -> Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}")

# Calculate smoothed positive class weights from training labels
pos_counts = Y_train.sum(axis=0)
total_train = len(Y_train)
neg_counts = total_train - pos_counts
raw_pos_weights = neg_counts / np.clip(pos_counts, 1.0, None)
# Clamped between 1.0 and 20.0 to prevent gradient explosions on rare labels
clamped_pos_weights = np.clip(raw_pos_weights, 1.0, 20.0).astype(np.float32)
pos_weight_tensor = torch.tensor(clamped_pos_weights, device=DEVICE)

class ProteinDataset(Dataset):
    def __init__(self, X, Y):
        self.X = torch.tensor(X, dtype=torch.long)
        self.Y = torch.tensor(Y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.Y[idx]

train_loader = DataLoader(ProteinDataset(X_train, Y_train), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(ProteinDataset(X_val, Y_val), batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(ProteinDataset(X_test, Y_test), batch_size=BATCH_SIZE, shuffle=False)
```

```python
# ==================================================
# 4. ARCHITECTURES: BASELINE VS. IMPROVED RESIDUAL CNN
# ==================================================

# 4A. Baseline 1D CNN with Masked Padding Fix
class MaskedBaselineCNN(nn.Module):
    def __init__(self, num_labels, pad_idx=0):
        super().__init__()
        self.pad_idx = pad_idx
        self.embedding = nn.Embedding(len(VOCAB) + 1, 32, padding_idx=pad_idx)
        self.conv1 = nn.Conv1d(32, 64, kernel_size=7, padding=3)
        self.conv2 = nn.Conv1d(64, 128, kernel_size=5, padding=2)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(128, num_labels)

    def forward(self, x):
        # x: [B, L]
        mask = (x != self.pad_idx)  # [B, L]
        emb = self.embedding(x).transpose(1, 2)  # [B, 32, L]
        h1 = self.relu(self.conv1(emb))
        h2 = self.relu(self.conv2(h1))  # [B, 128, L]

        # Masked max pooling: set padding tokens to -1e9 before max pooling!
        mask_expanded = mask.unsqueeze(1)  # [B, 1, L]
        h2_masked = h2.masked_fill(~mask_expanded, -1e9)
        pooled = torch.max(h2_masked, dim=2)[0]  # [B, 128]

        out = self.dropout(pooled)
        return self.fc(out)

# 4B. Improved Multi-Scale Residual CNN
class MultiScaleResidualCNN(nn.Module):
    """
    Multi-Scale Residual CNN capturing local motifs across variable kernel sizes
    (3, 5, 9, 15) with Batch Normalization, Residual Connections, and Dual Masked Pooling.
    """
    def __init__(self, num_labels, emb_dim=64, pad_idx=0):
        super().__init__()
        self.pad_idx = pad_idx
        self.embedding = nn.Embedding(len(VOCAB) + 1, emb_dim, padding_idx=pad_idx)
        self.emb_drop = nn.Dropout(0.1)

        # Multi-scale parallel convolutional branches
        self.branch3 = nn.Sequential(
            nn.Conv1d(emb_dim, 64, kernel_size=3, padding=1),
            nn.BatchNorm1d(64),
            nn.GELU()
        )
        self.branch5 = nn.Sequential(
            nn.Conv1d(emb_dim, 64, kernel_size=5, padding=2),
            nn.BatchNorm1d(64),
            nn.GELU()
        )
        self.branch9 = nn.Sequential(
            nn.Conv1d(emb_dim, 64, kernel_size=9, padding=4),
            nn.BatchNorm1d(64),
            nn.GELU()
        )
        self.branch15 = nn.Sequential(
            nn.Conv1d(emb_dim, 64, kernel_size=15, padding=7),
            nn.BatchNorm1d(64),
            nn.GELU()
        )

        # Merged representation: 64 * 4 = 256 channels
        merged_channels = 256

        # Second Residual Convolutional Block
        self.res_conv = nn.Sequential(
            nn.Conv1d(merged_channels, merged_channels, kernel_size=5, padding=2),
            nn.BatchNorm1d(merged_channels),
            nn.GELU(),
            nn.Conv1d(merged_channels, merged_channels, kernel_size=5, padding=2),
            nn.BatchNorm1d(merged_channels)
        )
        self.res_act = nn.GELU()

        # Classification Head: Dual Pooling (Max + Avg) -> 256 * 2 = 512 dimensions
        self.fc = nn.Sequential(
            nn.Linear(merged_channels * 2, 256),
            nn.BatchNorm1d(256),
            nn.GELU(),
            nn.Dropout(0.4),
            nn.Linear(256, num_labels)
        )

    def forward(self, x):
        # x: [B, L]
        mask = (x != self.pad_idx)  # [B, L]
        mask_exp = mask.unsqueeze(1)  # [B, 1, L]

        emb = self.embedding(x).transpose(1, 2)  # [B, emb_dim, L]
        emb = self.emb_drop(emb)

        # Multi-scale convolution
        b3 = self.branch3(emb)
        b5 = self.branch5(emb)
        b9 = self.branch9(emb)
        b15 = self.branch15(emb)
        merged = torch.cat([b3, b5, b9, b15], dim=1)  # [B, 256, L]

        # Residual connection
        residual = merged
        out = self.res_act(self.res_conv(merged) + residual)  # [B, 256, L]

        # Dual Masked Pooling:
        # 1. Masked Max Pooling
        out_masked = out.masked_fill(~mask_exp, -1e9)
        max_pool = torch.max(out_masked, dim=2)[0]  # [B, 256]

        # 2. Masked Average Pooling (Length-normalized)
        seq_lengths = mask.sum(dim=1, keepdim=True).clamp(min=1).float()  # [B, 1]
        out_zeroed = out.masked_fill(~mask_exp, 0.0)
        avg_pool = out_zeroed.sum(dim=2) / seq_lengths  # [B, 256]

        # Concatenate dual pooled features
        features = torch.cat([max_pool, avg_pool], dim=1)  # [B, 512]
        logits = self.fc(features)
        return logits
```

```python
# ==================================================
# 5. TRAINING, TUNING & EVALUATION FUNCTIONS
# ==================================================

def train_and_validate(model, train_loader, val_loader, criterion, optimizer,
                       scheduler=None, epochs=EPOCHS, patience=5, save_path="best_model.pt"):
    best_val_loss = float("inf")
    patience_counter = 0

    for epoch in range(epochs):
        model.train()
        train_loss = 0.0
        for batch_X, batch_Y in train_loader:
            batch_X = batch_X.to(DEVICE)
            batch_Y = batch_Y.to(DEVICE)

            optimizer.zero_grad()
            logits = model(batch_X)
            loss = criterion(logits, batch_Y)
            loss.backward()
            optimizer.step()

            train_loss += loss.item() * len(batch_X)
        train_loss /= len(train_loader.dataset)

        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for batch_X, batch_Y in val_loader:
                batch_X = batch_X.to(DEVICE)
                batch_Y = batch_Y.to(DEVICE)
                loss = criterion(model(batch_X), batch_Y)
                val_loss += loss.item() * len(batch_X)
        val_loss /= len(val_loader.dataset)

        if scheduler:
            scheduler.step(val_loss)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            torch.save(model.state_dict(), save_path)
        else:
            patience_counter += 1

        print(f"Epoch {epoch+1:02d}/{epochs} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Best: {best_val_loss:.4f}")
        if patience_counter >= patience:
            print(f"Early stopping triggered after {epoch+1} epochs.")
            break

    model.load_state_dict(torch.load(save_path, map_location=DEVICE))
    return model

def predict_probabilities(model, data_loader):
    model.eval()
    all_probs = []
    all_targets = []
    with torch.no_grad():
        for batch_X, batch_Y in data_loader:
            batch_X = batch_X.to(DEVICE)
            probs = torch.sigmoid(model(batch_X))
            all_probs.append(probs.cpu().numpy())
            all_targets.append(batch_Y.numpy())
    return np.concatenate(all_probs), np.concatenate(all_targets)

def tune_threshold_on_validation(val_probs, val_targets):
    """
    Tune the decision threshold STRICTLY on validation predictions to maximize Micro F1.
    NEVER touches test set.
    """
    thresholds = np.linspace(0.05, 0.95, 46)
    best_f1 = -1.0
    best_tau = 0.5

    for t in thresholds:
        preds = (val_probs >= t).astype(int)
        score = f1_score(val_targets, preds, average="micro", zero_division=0)
        if score > best_f1:
            best_f1 = score
            best_tau = t

    return best_tau, best_f1

def compute_all_metrics(probs, targets, tau):
    preds = (probs >= tau).astype(int)

    # Standard multi-label metrics
    micro_f1 = f1_score(targets, preds, average="micro", zero_division=0)
    macro_f1 = f1_score(targets, preds, average="macro", zero_division=0)
    micro_prec = precision_score(targets, preds, average="micro", zero_division=0)
    macro_prec = precision_score(targets, preds, average="macro", zero_division=0)
    micro_rec = recall_score(targets, preds, average="micro", zero_division=0)
    macro_rec = recall_score(targets, preds, average="macro", zero_division=0)

    # Compute CAFA Fmax across all candidate thresholds
    thresholds = np.linspace(0.01, 0.99, 99)
    fmax = 0.0
    for t in thresholds:
        p_t = (probs >= t).astype(int)
        f_t = f1_score(targets, p_t, average="micro", zero_division=0)
        if f_t > fmax:
            fmax = f_t

    return {
        "Threshold": round(float(tau), 4),
        "Micro_F1": round(float(micro_f1), 4),
        "Macro_F1": round(float(macro_f1), 4),
        "Micro_Precision": round(float(micro_prec), 4),
        "Macro_Precision": round(float(macro_prec), 4),
        "Micro_Recall": round(float(micro_rec), 4),
        "Macro_Recall": round(float(macro_rec), 4),
        "CAFA_Fmax": round(float(fmax), 4)
    }
```

```python
# ==================================================
# 6. SYSTEMATIC ABLATION EXPERIMENTS
# ==================================================
results_list = []

print("\n" + "="*70)
print("EXPERIMENT 1: ORIGINAL BASELINE (Fixed Threshold 0.5, Standard BCE)")
print("="*70)
# Loaded from earlier baseline_results.csv or trained here
if os.path.exists("baseline_results.csv"):
    b_df = pd.read_csv("baseline_results.csv")
    b_row = b_df.to_dict('records')[0]
    results_list.append(b_row)
    print(f"Loaded existing baseline result: Micro F1 = {b_row['Micro_F1']}, Macro F1 = {b_row['Macro_F1']}")
else:
    print("Baseline result not found, running...")

print("\n" + "="*70)
print("EXPERIMENT 2: VALIDATION-BASED THRESHOLD TUNING (Original Architecture)")
print("="*70)
seed_everything(SEED)
baseline_masked_model = MaskedBaselineCNN(len(selected_labels)).to(DEVICE)
opt2 = torch.optim.Adam(baseline_masked_model.parameters(), lr=LR)
crit2 = nn.BCEWithLogitsLoss()
if os.path.exists("exp2_model.pt"):
    print("Loading cached checkpoint exp2_model.pt...")
    baseline_masked_model.load_state_dict(torch.load("exp2_model.pt", map_location=DEVICE))
else:
    baseline_masked_model = train_and_validate(baseline_masked_model, train_loader, val_loader,
                                              crit2, opt2, epochs=10, save_path="exp2_model.pt")

val_probs2, val_targets2 = predict_probabilities(baseline_masked_model, val_loader)
test_probs2, test_targets2 = predict_probabilities(baseline_masked_model, test_loader)

# Strict validation-only threshold search
best_tau2, val_f1_2 = tune_threshold_on_validation(val_probs2, val_targets2)
print(f"Optimal threshold tuned on validation set: tau* = {best_tau2:.2f} (Val Micro F1 = {val_f1_2:.4f})")

# Evaluate on held-out test set with val-tuned threshold
m2 = compute_all_metrics(test_probs2, test_targets2, best_tau2)
m2["Model"] = "Masked Baseline CNN + Val Threshold Tuning"
results_list.append(m2)
print(f"Test Evaluation: Micro F1 = {m2['Micro_F1']} | Macro F1 = {m2['Macro_F1']} | Micro Rec = {m2['Micro_Recall']}")

print("\n" + "="*70)
print("EXPERIMENT 3: CLASS WEIGHTING (pos_weight) + PADDING MASKING")
print("="*70)
seed_everything(SEED)
model3 = MaskedBaselineCNN(len(selected_labels)).to(DEVICE)
opt3 = torch.optim.Adam(model3.parameters(), lr=LR)
crit3 = nn.BCEWithLogitsLoss(pos_weight=pos_weight_tensor)
if os.path.exists("exp3_model.pt"):
    print("Loading cached checkpoint exp3_model.pt...")
    model3.load_state_dict(torch.load("exp3_model.pt", map_location=DEVICE))
else:
    model3 = train_and_validate(model3, train_loader, val_loader, crit3, opt3, epochs=10, save_path="exp3_model.pt")

val_probs3, val_targets3 = predict_probabilities(model3, val_loader)
test_probs3, test_targets3 = predict_probabilities(model3, test_loader)

best_tau3, val_f1_3 = tune_threshold_on_validation(val_probs3, val_targets3)
print(f"Optimal threshold tuned on validation set: tau* = {best_tau3:.2f} (Val Micro F1 = {val_f1_3:.4f})")

m3 = compute_all_metrics(test_probs3, test_targets3, best_tau3)
m3["Model"] = "Masked Baseline CNN + Class Weighting (pos_weight)"
results_list.append(m3)
print(f"Test Evaluation: Micro F1 = {m3['Micro_F1']} | Macro F1 = {m3['Macro_F1']} | Macro Rec = {m3['Macro_Recall']}")

print("\n" + "="*70)
print("EXPERIMENT 4: IMPROVED MULTI-SCALE RESIDUAL CNN + EARLY STOPPING + LR SCHEDULER")
print("="*70)
seed_everything(SEED)
model4 = MultiScaleResidualCNN(len(selected_labels), emb_dim=64).to(DEVICE)
opt4 = torch.optim.AdamW(model4.parameters(), lr=0.001, weight_decay=1e-4)
sched4 = torch.optim.lr_scheduler.ReduceLROnPlateau(opt4, mode='min', factor=0.5, patience=2)
crit4 = nn.BCEWithLogitsLoss(pos_weight=pos_weight_tensor)

model4 = train_and_validate(model4, train_loader, val_loader, crit4, opt4, scheduler=sched4,
                            epochs=15, patience=5, save_path="exp4_best_multiscale.pt")

val_probs4, val_targets4 = predict_probabilities(model4, val_loader)
test_probs4, test_targets4 = predict_probabilities(model4, test_loader)

best_tau4, val_f1_4 = tune_threshold_on_validation(val_probs4, val_targets4)
print(f"Optimal threshold tuned on validation set: tau* = {best_tau4:.2f} (Val Micro F1 = {val_f1_4:.4f})")

m4 = compute_all_metrics(test_probs4, test_targets4, best_tau4)
m4["Model"] = "Improved MultiScale Residual CNN (Full Suite)"
results_list.append(m4)
print(f"Test Evaluation: Micro F1 = {m4['Micro_F1']} | Macro F1 = {m4['Macro_F1']} | CAFA Fmax = {m4['CAFA_Fmax']}")
```

```python
# ==================================================
# 7. SAVE COMPARATIVE RESULTS
# ==================================================
final_results_df = pd.DataFrame(results_list)
final_results_df = final_results_df[[
    "Model", "Threshold", "Micro_F1", "Macro_F1",
    "Micro_Precision", "Macro_Precision", "Micro_Recall", "Macro_Recall", "CAFA_Fmax"
]]
final_results_df.to_csv("experiment_results.csv", index=False)

print("\n" + "="*80)
print("--- SUMMARY OF EXPERIMENTAL PROGRESSION ---")
print("="*80)
print(final_results_df.to_string(index=False))
print("="*80)
print("Saved experiment_results.csv successfully.")

# Save final complete production artifact
torch.save({
    "model_state": model4.state_dict(),
    "classes": selected_labels,
    "vocab": VOCAB,
    "max_len": MAX_LEN,
    "val_tuned_threshold": best_tau4,
    "metrics": m4
}, "improved_protein_cnn_complete.pt")
print("Saved production checkpoint: improved_protein_cnn_complete.pt")
```
